```python
# Define the features to be used for clustering
cluster_features = ['TB', 'DB', 'Alkphos', 'Sgpt', 'Sgot', 'TP', 'ALB', 'A/G Ratio']
```

```python
# Create a new DataFrame containing only the selected features for clustering
x = df[cluster_features].copy()
x.head()
```

```python
# Display the shape of the new DataFrame containing only the selected features for clustering
x.shape
```

```python
# Apply log transformation to skewed features to reduce skewness and improve clustering performance
log_features = ['TB', 'DB', 'Alkphos', 'Sgpt', 'Sgot', 'A/G Ratio']
x_transformed = x.copy()

for feature in log_features:
    x_transformed[feature] = np.log1p(x_transformed[feature])
```

```python
# Display the skewness of the transformed features to verify the effect of log transformation
x_transformed[log_features].skew()
```

```python
# Standardize the features to have a mean of 0 and a standard deviation of 1
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x_transformed)

x_scaled.shape
```

```python
# Display the standard deviation of the scaled features to verify that they have been standardized
x_scaled.mean(axis=0)
```

```python
# Display the standard deviation of the scaled features to verify that they have been standardized
x_scaled.std(axis=0)
```

```python
# Use the elbow method to determine the optimal number of clusters for KMeans clustering
from sklearn.cluster import KMeans

inertia = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(x_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Determining Optimal Number of Clusters')
plt.show()
```

```python
# Use silhouette scores to determine the optimal number of clusters for KMeans clustering
from sklearn.metrics import silhouette_score

silhouette_scores = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(x_scaled)
    score = silhouette_score(x_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Scores by Number of Clusters')
plt.show()
```

```python
# Perform KMeans clustering with the optimal number of clusters (k=2)
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

cluster_labels = kmeans.fit_predict(x_scaled)
```

```python
# Create a new DataFrame to store the original data along with the cluster labels
df_clustered = df.copy()
df_clustered['Cluster'] = cluster_labels
```

```python
# Display the number of patients in each cluster
df_clustered['Cluster'].value_counts().sort_index()
```

```python
# Display the mean values of the features for each cluster to understand the characteristics of each cluster
cluster_profiles = df_clustered.groupby('Cluster')[cluster_features].mean()
cluster_profiles
```

```python
# Display the distribution of age across clusters
df_clustered.groupby('Cluster')['Age'].mean()
```

```python
# Display the distribution of gender across
pd.crosstab(df_clustered['Cluster'], df_clustered['Gender'], normalize='index') * 100
```

```python
# Perform PCA to reduce the dimensionality of the data for visualization
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

x_pca = pca.fit_transform(x_scaled)

pca_df = pd.DataFrame(data=x_pca, columns=['PCA1', 'PCA2'])
pca_df['Cluster'] = cluster_labels
pca_df.head()
```

```python
# Visualize the clusters in the PCA-reduced space
plt.figure(figsize=(8, 6))

for cluster in sorted(pca_df['Cluster'].unique()):
    cluster_data = pca_df[pca_df['Cluster'] == cluster]
    plt.scatter(cluster_data['PCA1'], cluster_data['PCA2'], label=f'Cluster {cluster}', alpha=0.7)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Patient Clusters Visualized Using PCA')
plt.legend()
plt.show()
```

```python
# Display the explained variance ratio of the PCA components to understand how much variance is captured by each component
print('Explained variance ratio:')
print(pca.explained_variance_ratio_)
```

```python
# Display the loadings of the PCA components to understand the contribution of each feature to the principal components
loadings = pd.DataFrame(pca.components_.T, columns=['PC1', 'PC2'], index=cluster_features)
loadings
```

```python
# Evaluate the clustering performance using Calinski-Harabasz and Davies-Bouldin scores
from sklearn.metrics import calinski_harabasz_score, davies_bouldin_score

calinski_score = calinski_harabasz_score(x_scaled, cluster_labels)
davies_score = davies_bouldin_score(x_scaled, cluster_labels)

print(f'Calinski-Harabasz Score: {calinski_score}')
print(f'Davies-Bouldin Score: {davies_score}')
```

```python
# Create a summary of the clusters, including the number of patients and mean values of each feature for each cluster
cluster_summary = df_clustered.groupby('Cluster').agg(
    patient_count=('Cluster', 'size'),
    Mean_Age=('Age', 'mean'),
    Mean_TB=('TB', 'mean'),
    Mean_DB=('DB', 'mean'),
    Mean_Alkphos=('Alkphos', 'mean'),
    Mean_Sgpt=('Sgpt', 'mean'),
    Mean_Sgot=('Sgot', 'mean'),
    Mean_TP=('TP', 'mean'),
    Mean_ALB=('ALB', 'mean'),
    Mean_AG_Ratio=('A/G Ratio', 'mean')
)

cluster_summary
```
